```python
# Calculate Rate of Return

# S&P 500 returns since year 2000
import pandas as pd, numpy as np, matplotlib.pyplot as plt, yfinance as yf

raw_data = yf.download('^NSEI', start='2000-1-1')
data = pd.DataFrame(raw_data)
data.rename(columns={'^NSEI' : 'Nifty50'}, inplace = True)
print(data)
```

```python
plt.plot(data['Close'])
plt.show()
```

```python

# Calculate simple rate of return


ret = data['Close'].pct_change() #  calculate daily % change 

# calculate annualize mean return
annual_avg_ret = round(ret.mean()*250*100,3) # finds the daily avg return of last 25 years and annualize it

print(annual_avg_ret) 
# this method is suitable for short term estimate - but can be inaccurate due to daily fluctuations 
```

```python

plt.plot(ret) 
plt.title('Nifty50 Daily % change since 2000')
plt.show()
```

```python
# calculate CAGR in %

ret2 = ((data['Close'].iloc[-1]/data['Close'].iloc[0])**(1/24) - 1)*100 # takes the starting and closing price and find geometric return
print(ret2) # used for long term tenure returns
```

```python

# Calculate logarthemic rate of return

log_ret = np.log(data['Close']/data['Close'].shift(1))
print(log_ret)
```

```python
# mean return
log_ret_avg = log_ret.mean()
print(log_ret_avg)
```

```python
# annualize the return
annual_log_ret_avg = round(log_ret_avg*250*100,3) # annualizing log returns
print(annual_log_ret_avg)
```

```python
plt.plot(log_ret)
plt.show()
```
